```python
from langgraph.graph import StateGraph, START, END
from langchain_openai import ChatOpenAI
from typing import TypedDict
from dotenv import load_dotenv
import os
from getpass import getpass
```

```python
load_dotenv()
```

```python
os.environ["OPENAI_API_KEY"] = getpass("OpenAI API key: ")
```

```python
model = ChatOpenAI(model="gpt-5-nano")
```

```python
class LLMState(TypedDict):
    question: str
    answer: str
```

```python
def llm_query(state: LLMState) -> LLMState:

    question = state["question"]

    prompt = f"Answer the following question: {question}"

    answer = model.invoke(prompt).content

    state["answer"] = answer
    return state
```

```python
graph = StateGraph(LLMState)

# nodes
graph.add_node("llm_query", llm_query)

# edges
graph.add_edge(START, "llm_query")
graph.add_edge("llm_query", END)

workflow = graph.compile()
```

```python
# execute

initialState: LLMState = {"question": "Which is faster me or sound of light"}

finalState: LLMState = workflow.invoke(initialState)

print(finalState["answer"])
```
